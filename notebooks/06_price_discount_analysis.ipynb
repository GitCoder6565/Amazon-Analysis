# 💸 Price & Discount Analysis

This notebook implements the sixth dimension of our analytical framework: **Price & Discount Analysis**.

**Objective:** Quantifying price elasticity and identifying 'Profit Leaks' from over-discounting.

## 1. Setup and Data Loading

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style
plt.style.use('ggplot')

# Load data
df = pd.read_csv('../Amazon.csv')
success_orders = df[df['OrderStatus'].isin(['Delivered', 'Shipped'])].copy()
```

## 2. Price Elasticity of Demand (PED)
Analyzing how changes in Unit Price affect the Quantity sold.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='UnitPrice', y='Quantity', data=success_orders, alpha=0.5)
plt.title('Relationship between Unit Price and Quantity Sold', fontweight='bold')
plt.show()

corr = success_orders[['UnitPrice', 'Quantity']].corr().iloc[0,1]
print(f"Correlation between UnitPrice and Quantity: {corr:.4f}")
```

## 3. Discount Efficiency
Total revenue lost to discounts versus the gain in volume.

```python
# Calculate Gross Revenue (before discount)
success_orders['GrossRevenue'] = success_orders['Quantity'] * success_orders['UnitPrice']
success_orders['DiscountAmount'] = success_orders['GrossRevenue'] * success_orders['Discount']

total_gross = success_orders['GrossRevenue'].sum()
total_discount = success_orders['DiscountAmount'].sum()

print(f"Total Potential Gross Revenue: ${total_gross:,.2f}")
print(f"Total Revenue Lost to Discounts: ${total_discount:,.2f}")
print(f"Overall Discount Impact: {(total_discount / total_gross)*100:.2f}%")
```

## 4. Discount Impact by Category
Which categories are most 'Discount Sensitive'?

```python
cat_discount = success_orders.groupby('Category')['Discount'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
cat_discount.plot(kind='bar', color='tomato')
plt.title('Average Discount level by Category', fontweight='bold')
plt.ylabel('Average Discount (%)')
plt.xticks(rotation=45)
plt.show()
```
