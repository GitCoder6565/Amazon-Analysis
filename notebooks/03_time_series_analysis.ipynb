# 📈 Time-Series Analysis: Seasonality and Trends

This notebook implements the third dimension of our analytical framework: **Time-Series Analysis**.

**Objective:** Identifying when sales happen, detecting seasonality, and analyzing consumer behavior cycles (e.g., weekends vs. weekdays).

## 1. Setup and Data Loading

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('ggplot')
sns.set_palette("viridis")

# Load data
df = pd.read_csv('../Amazon.csv')

# Convert OrderDate to datetime
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Filter for successful orders
success_orders = df[df['OrderStatus'].isin(['Delivered', 'Shipped'])].copy()
```

## 2. Monthly Sales Trends
How has revenue evolved over the months?

```python
# Aggregate monthly revenue
monthly_revenue = success_orders.resample('M', on='OrderDate')['TotalAmount'].sum().reset_index()
monthly_revenue['Month-Year'] = monthly_revenue['OrderDate'].dt.strftime('%b %Y')

plt.figure(figsize=(15, 6))
sns.lineplot(data=monthly_revenue, x='OrderDate', y='TotalAmount', marker='o', color='#2ca02c')
plt.title('Monthly Revenue Trend', fontweight='bold', fontsize=15)
plt.ylabel('Total Revenue ($)')
plt.xlabel('Date')
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Seasonality Analysis (Monthly Averages)
Are there specific months that consistently perform better?

```python
success_orders['Month'] = success_orders['OrderDate'].dt.month_name()
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']

avg_monthly_rev = success_orders.groupby('Month')['TotalAmount'].mean().reindex(month_order)

plt.figure(figsize=(12, 6))
avg_monthly_rev.plot(kind='bar', color='skyblue')
plt.title('Average Revenue by Month (Seasonality)', fontweight='bold')
plt.ylabel('Average Revenue ($)')
plt.xticks(rotation=45)
plt.show()
```

## 4. Cycle Analysis: Weekday vs. Weekend
Do consumers shop more on weekends?

```python
success_orders['DayType'] = success_orders['OrderDate'].dt.dayofweek.apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')
day_type_revenue = success_orders.groupby('DayType')['TotalAmount'].sum()

plt.figure(figsize=(8, 6))
day_type_revenue.plot(kind='pie', autopct='%1.1f%%', colors=['#ff7f0e', '#1f77b4'], startangle=90)
plt.title('Revenue Contribution: Weekday vs. Weekend', fontweight='bold')
plt.ylabel('')
plt.show()
```
