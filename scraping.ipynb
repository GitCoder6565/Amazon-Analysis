```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://books.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
books = soup.find_all('article', class_ = 'product_pod')
data = []
for book in books:
    data.append(
        {
            'name': book.find('h3').find('a')['title'],
            'price': book.find('p', class_ = 'price_color').text.replace('£', ''),
            'rating': book.find('p', class_='star-rating').get('class')[1]
        }
    )
```

```python
pd.set_option('display.max_colwidth', None)
df = pd.DataFrame(data)
df
```

```python
# 1. Create a dictionary to translate words to numbers
rating_map = {
    'One': 1,
    'Two': 2,
    'Three': 3,
    'Four': 4,
    'Five': 5
}

# 2. Extract the class name (e.g., "Three") and map it
# Note: This assumes you use the .get('class') method mentioned previously
df['rating'] = df['rating (out of 5 stars)'].map(rating_map)
df.drop(columns=['rating (out of 5 stars)'], inplace=True)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "http://books.toscrape.com/catalogue/page-1.html"
all_data = []

while url:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # --- 1. Your Existing Scraping Logic ---
    books = soup.find_all('article', class_='product_pod')
    for book in books:
        all_data.append({
            'name': book.find('h3').find('a')['title'],
            'price': book.find('p', class_='price_color').text.replace('Â', ''),
            'rating': book.find('p', class_='star-rating').get('class')[1]
        })

    # --- 2. Finding the "Next" Button ---
    next_btn = soup.find('li', class_='next')
    
    if next_btn:
        next_url = next_btn.find('a')['href']
        # This site uses relative URLs, so we need to join them
        from urllib.parse import urljoin
        url = urljoin(url, next_url)
    else:
        url = None  # No more pages, exit the loop

df = pd.DataFrame(all_data)
```

```python
df.to_csv('books.csv', index=False)
from pandas import to_excel
df.to_excel('books.xlsx', index=False)
```
